```python
import torch
import torchvision

from torchvision import datasets, transforms

import matplotlib.pyplot as plt
```

```python
print("PyTorch version:", torch.__version__)
print("Torchvision version:", torchvision.__version__)
```

```python
transform = transforms.Compose([
    transforms.ToTensor()
])
```

```python
train_dataset = datasets.CIFAR10(
    root="./data",
    train=True,
    download=True,
    transform=transform
)
```

```python
test_dataset = datasets.CIFAR10(
    root="./data",
    train=False,
    download=True,
    transform=transform
)
```

```python
print("Training images:", len(train_dataset))
print("Testing images:", len(test_dataset))
```

```python
print(train_dataset.classes)
```

```python
image, label = train_dataset[0]

print("Label number:", label)
print("Class name:", train_dataset.classes[label])
```

```python
plt.imshow(image.permute(1, 2, 0))
plt.title(train_dataset.classes[label])
plt.axis("off")
plt.show()
```

```python
known_classes = [
    "airplane",
    "automobile",
    "cat",
    "dog",
    "horse"
]

unknown_classes = [
    "bird",
    "deer",
    "frog",
    "ship",
    "truck"
]
```

```python
class_to_idx = train_dataset.class_to_idx

known_indices = [
    class_to_idx[class_name]
    for class_name in known_classes
]

unknown_indices = [
    class_to_idx[class_name]
    for class_name in unknown_classes
]

print("Known indices:", known_indices)
print("Unknown indices:", unknown_indices)
```

```python
known_train_indices = []

for i, (_, label) in enumerate(train_dataset):
    if label in known_indices:
        known_train_indices.append(i)

print("Known training samples:", len(known_train_indices))
```
